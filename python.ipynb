```python


import pandas as pd
import os

# Step 1: Load dataset
file_path = r'C:\Users\Admin\Downloads\customer_shopping_behavior.csv'
df = pd.read_csv(file_path)

print("Original Dataset Shape:", df.shape)

# Step 2: Check missing values before cleaning
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# Step 3: Fill missing Review Ratings using category median
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(
    lambda x: x.fillna(x.median())
)

# Step 4: Fill any remaining missing values with 0
df = df.fillna(0)

# Step 5: Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# Step 6: Calculate total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())

# Step 7: Save cleaned dataset
output_path = r'C:\Users\Admin\Downloads\customer_shopping_behavior_cleaned.csv'
df.to_csv(output_path, index=False)

print("\nCleaned Dataset Saved Successfully!")

# Step 8: Display cleaned dataset
display(df.head())
```

```python

import pandas as pd
import os

# Load CSV file using its complete path
file_path = r'C:\Users\Admin\Downloads\customer_shopping_behavior.csv'

df = pd.read_csv(file_path)

# Remove unwanted spaces from column names
df.columns = df.columns.str.strip()

# Define age group labels
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

# Create age groups
df['age_group'] = pd.qcut(
    df['Age'],
    q=4,
    labels=labels
)

# Display first 10 rows
print("Age Groups Created Successfully!")

display(df[['Age', 'age_group']].head(10))
```

```python

import pandas as pd

# Load dataset
df = pd.read_csv(
    r'C:\Users\Admin\Downloads\customer_shopping_behavior.csv'
)

# Clean column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# Check column names
print(df.columns.tolist())

# Create frequency mapping
frequency_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Bi-Weekly': 14,
    'Annually': 365
}

# Convert purchase frequency into days
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(
    frequency_mapping
)

# Display results
display(df[['frequency_of_purchases', 'purchase_frequency_days']].head(10))
```
